# 🧹 Data Cleaning

## Project Overview

The dataset is already well-structured with no missing values or duplicate records. Therefore, this notebook focuses on validating data quality, standardizing the dataset, optimizing data types, and preparing the data for feature engineering.

## Objectives

- Convert columns to appropriate data types
- Validate numerical columns
- Check categorical consistency
- Remove redundant information
- Standardize column names
- Prepare a clean dataset for feature engineering

```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
pd.set_option("display.float_format", "{:.2f}".format)
```

```python
# Load dataset

df = pd.read_csv("../data/raw/digital_marketing_dataset_30k.csv")

df.head()
```

```python
print(f"Rows    : {df.shape[0]}")
print(f"Columns : {df.shape[1]}")
```

```python
# Check current datatype

df["date"].dtype
```

```python
df["date"] = pd.to_datetime(
    df["date"],
    format="%d/%m/%Y",
    dayfirst=True
)
```

```python
df["date"].dtype
```

```python
df.isnull().sum().sum()
```

```python
df.duplicated().sum()
```

```python
numerical_cols = [
    "spend",
    "impressions",
    "reach",
    "frequency",
    "clicks",
    "conversions",
    "revenue",
    "video_views"
]

df[numerical_cols].describe()
```

```python
(df[numerical_cols] < 0).sum()
```

```python
pd.crosstab(df["is_holiday"], df["is_holiday (text)"])
```

```python
pd.crosstab(df["is_weekend"], df["is_weekend (text)"])
```

```python
df.drop(
    columns=[
        "is_holiday (text)",
        "is_weekend (text)"
    ],
    inplace=True
)
```

```python
# Verify updated dataset shape

print(f"Rows    : {df.shape[0]}")
print(f"Columns : {df.shape[1]}")
```

```python
df.columns = (
    df.columns
      .str.lower()
      .str.replace(" ", "_")
      .str.replace("(", "", regex=False)
      .str.replace(")", "", regex=False)
)
```

```python
df.columns
```

```python
df.info()
```

```python
df.to_csv("../data/processed/marketing_campaign_cleaned.csv", index=False)

print("Cleaned dataset saved successfully!")
```

# Key Findings

- Converted the `date` column from object to datetime format.
- Verified that the dataset contains no missing values.
- Confirmed that no duplicate records are present.
- Validated that numerical columns contain no negative values.
- Verified consistency between numeric and text versions of holiday and weekend indicators.
- Removed redundant columns:
  - is_holiday (text)
  - is_weekend (text)
- Standardized column names using snake_case.
- Exported the cleaned dataset for feature engineering.
