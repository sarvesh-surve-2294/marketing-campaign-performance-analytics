# 🗄️ Load Data into MySQL

## Project Overview

This notebook prepares the feature-engineered dataset for SQL analysis by loading it into a MySQL database.

The dataset is exported as a CSV file and then imported into MySQL, where SQL queries will be executed to analyze campaign performance and generate business insights.

## Objectives

- Load the feature-engineered dataset
- Verify the dataset structure
- Export the final dataset for MySQL import
- Validate the exported file.

```python
import pandas as pd

pd.set_option("display.max_columns", None)
```

```python
# Load dataset

df = pd.read_csv(
    "../data/processed/marketing_campaign_feature_engineered.csv"
)

df.head()
```

```python
print(f"Rows    : {df.shape[0]}")
print(f"Columns : {df.shape[1]}")

df.info()
```

```python
# Export final dataset for MySQL

output_path = "../data/processed/marketing_campaign_mysql.csv"

df.to_csv(output_path, index=False)

print(f"Dataset exported successfully to:\n{output_path}")
```

```python
# Read exported file

mysql_df = pd.read_csv(output_path)

print(f"Rows    : {mysql_df.shape[0]}")
print(f"Columns : {mysql_df.shape[1]}")

mysql_df.head()
```

# Key Findings

- Successfully loaded the feature-engineered dataset.
- Verified dataset dimensions and column structure.
- Exported the final dataset for MySQL import.
- Validated the exported CSV file to ensure data integrity.
