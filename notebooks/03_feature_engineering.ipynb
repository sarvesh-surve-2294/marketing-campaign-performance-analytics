# 📈 Feature Engineering

## Project Overview

This notebook focuses on creating new marketing KPIs and derived features from the cleaned dataset. These engineered features will be used for SQL analysis, Power BI dashboards, and business insights.

## Objectives

- Create marketing performance metrics
- Handle division-by-zero scenarios
- Prepare the dataset for exploratory analysis
- Export the enriched dataset

```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
pd.set_option("display.float_format", "{:.2f}".format)
```

```python
# Load cleaned dataset

df = pd.read_csv("../data/processed/marketing_campaign_cleaned.csv")

df.head()
```

```python
# Click Through Rate

df["ctr"] = np.where(
    df["impressions"] > 0,
    (df["clicks"] / df["impressions"]) * 100,
    0
)
```

```python
# Cost Per Click

df["cpc"] = np.where(
    df["clicks"] > 0,
    df["spend"] / df["clicks"],
    0
)
```

```python
# Cost Per Thousand Impressions

df["cpm"] = np.where(
    df["impressions"] > 0,
    (df["spend"] / df["impressions"]) * 1000,
    0
)
```

```python
# Cost Per Acquisition

df["cpa"] = np.where(
    df["conversions"] > 0,
    df["spend"] / df["conversions"],
    0
)
```

```python
# Conversion Rate

df["conversion_rate"] = np.where(
    df["clicks"] > 0,
    (df["conversions"] / df["clicks"]) * 100,
    0
)
```

```python
# Return On Ad Spend

df["roas"] = np.where(
    df["spend"] > 0,
    df["revenue"] / df["spend"],
    0
)
```

```python
# Return On Investment

df["roi"] = np.where(
    df["spend"] > 0,
    ((df["revenue"] - df["spend"]) / df["spend"]) * 100,
    0
)
```

```python
# Revenue Per Click

df["revenue_per_click"] = np.where(
    df["clicks"] > 0,
    df["revenue"] / df["clicks"],
    0
)
```

```python
# Revenue Per Conversion

df["revenue_per_conversion"] = np.where(
    df["conversions"] > 0,
    df["revenue"] / df["conversions"],
    0
)
```

```python
kpi_columns = [
    "ctr",
    "cpc",
    "cpm",
    "cpa",
    "conversion_rate",
    "roas",
    "roi",
    "revenue_per_click",
    "revenue_per_conversion"
]

df[kpi_columns].describe()
```

- Safely handled division-by-zero scenarios while calculating KPIs.
